# Notebook 2 – Numerical Feature Engineering

## 1. What is Numerical Feature Engineering?

Numerical Feature Engineering is the process of creating new numerical features or transforming existing numerical features using mathematical operations and statistical techniques.

It helps Machine Learning models identify useful relationships and patterns in numerical data.

Example:

If a customer dataset contains Monthly Charges and Tenure, we can create a new feature called Total Spending.

Formula:

Total Spending = Monthly Charges × Tenure

## 2. Mathematical Transformations

Mathematical transformations use mathematical operations to create new features from existing numerical columns.

They help represent relationships between different variables.

### 2.1 Addition

Addition combines two or more numerical features.

Formula:

New Feature = Feature A + Feature B

Example:

Total Charges = Internet Charges + Phone Charges

When to use:

* Combining different expenses.
* Calculating total revenue.
* Combining multiple numerical measurements.

### 2.2 Subtraction

Subtraction calculates the difference between two numerical features.

Formula:

Difference = Feature A - Feature B

Example:

Remaining Balance = Total Amount - Paid Amount

When to use:

* Finding remaining balances.
* Calculating profit.
* Identifying differences between measurements.

### 2.3 Multiplication

Multiplication combines numerical features to represent their combined effect.

Formula:

New Feature = Feature A × Feature B

Example:

Total Spending = Monthly Charges × Tenure

When to use:

* Calculating total spending.
* Calculating revenue.
* Creating interaction features.

### 2.4 Division

Division calculates the relationship between two numerical features.

Formula:

New Feature = Feature A / Feature B

Example:

Average Spending = Total Spending / Number of Months

When to use:

* Calculating averages.
* Measuring efficiency.
* Calculating per-unit values.

Always handle division by zero to avoid infinity or invalid values.

## 3. Ratios

A ratio compares two numerical quantities.

Formula:

Ratio = Feature A / Feature B

Example:

Debt-to-Income Ratio = Total Debt / Annual Income

Why useful?

It helps compare values relative to each other instead of considering only their absolute amounts.

Applications:

* Banking
* Finance
* Customer spending analysis
* Business performance

## 4. Percentages

Percentage features express a value as a proportion of another value multiplied by 100.

Formula:

Percentage = (Part / Total) × 100

Example:

Discount Percentage = (Discount Amount / Original Price) × 100

Why useful?

Percentages make it easier to compare values across different scales.

## 5. Differences

Difference features calculate how much one numerical value differs from another.

Formula:

Difference = Current Value - Previous Value

Example:

Monthly Spending Change = Current Month Spending - Previous Month Spending

Why useful?

They help identify increases, decreases and changes in customer behaviour.

## 6. Aggregations

Aggregation combines multiple numerical values into summary measurements.

Common aggregation techniques:

* Sum
* Mean
* Median
* Minimum
* Maximum
* Count
* Standard Deviation

Example:

Total Customer Spending = Sum of all customer transactions.

Why useful?

Aggregation helps summarize large amounts of data into meaningful information.

## 7. Log Transformation

Log Transformation applies a logarithmic function to numerical values.

Example:

Log Feature = log(1 + Feature Value)

In Python:

np.log1p()

Why useful?

* Reduces right-skewness.
* Reduces the influence of very large values.
* Helps represent data with a wide range of values.
* Can improve performance for some Machine Learning algorithms.

It is particularly useful for highly skewed numerical features.

## 8. Square Root Transformation

Square Root Transformation calculates the square root of a numerical feature.

Formula:

New Feature = √Feature

In Python:

np.sqrt()

Why useful?

* Reduces positive skewness.
* Compresses large numerical values.
* Can make some distributions more balanced.

It is commonly used for non-negative numerical data.

## 9. Power Transformation

Power Transformation changes the distribution of numerical data using mathematical power functions.

Examples:

* Square Transformation: x²
* Cube Transformation: x³
* Box-Cox Transformation
* Yeo-Johnson Transformation

Why useful?

* Helps modify data distributions.
* Can reduce skewness.
* May improve the performance of scale-sensitive Machine Learning algorithms.

Box-Cox requires strictly positive values, whereas Yeo-Johnson can handle positive, zero and negative values.

## 10. Absolute Difference

Absolute Difference calculates the positive distance between two numerical values without considering their direction.

Formula:

Absolute Difference = |Feature A - Feature B|

Example:

Difference between Expected Spending and Actual Spending.

Why useful?

* Measures the size of an error.
* Identifies the magnitude of changes.
* Helps compare values without considering whether they increased or decreased.

## 11. Relative Difference

Relative Difference measures the difference between two values compared with a reference value.

Formula:

Relative Difference = (Current Value - Previous Value) / Previous Value

Why useful?

It measures proportional changes instead of absolute changes.

Example:

If spending increases from ₹1,000 to ₹1,200:

Relative Difference = (1200 - 1000) / 1000 = 0.20

Percentage Increase = 20%

Note: The reference value must not be zero.

## 12. Rate Features

Rate Features measure how quickly something changes or occurs over a specific period.

Formula:

Rate = Change in Value / Time Period

Example:

Monthly Purchase Rate = Number of Purchases / Number of Months

Why useful?

* Measures customer activity.
* Identifies transaction frequency.
* Helps understand business performance.
* Supports customer behaviour analysis.

## 13. Normalized Features

Normalization transforms numerical values into a common scale.

One common technique is Min-Max Normalization.

Formula:

X_normalized = (X - X_min) / (X_max - X_min)

The resulting values generally range between 0 and 1.

Why useful?

* Places numerical features on a common scale.
* Prevents large-scale numerical features from dominating distance-based calculations.
* Helps algorithms such as KNN and K-Means.

## 14. Importance of Numerical Feature Engineering

Numerical Feature Engineering helps:

1. Create meaningful information from existing numerical columns.
2. Identify relationships between variables.
3. Understand customer behaviour.
4. Reduce the effect of skewed distributions.
5. Improve the representation of numerical data.
6. Support better Machine Learning predictions.

## Conclusion

Numerical Feature Engineering converts existing numerical information into useful features through mathematical operations and transformations.

An AI/ML Engineer should understand the purpose of each transformation and select techniques based on the dataset, business problem and Machine Learning algorithm.

## 2.5 When Should We Use Numerical Transformations?

| Technique            | When to Use                             |
| -------------------- | --------------------------------------- |
| Addition             | Combining numerical values              |
| Subtraction          | Finding differences                     |
| Multiplication       | Calculating combined effects            |
| Division             | Calculating averages or per-unit values |
| Ratios               | Comparing two quantities                |
| Percentages          | Measuring proportional values           |
| Aggregations         | Summarizing multiple records            |
| Log Transformation   | Handling highly right-skewed data       |
| Square Root          | Reducing positive skewness              |
| Power Transformation | Modifying numerical distributions       |
| Absolute Difference  | Measuring the magnitude of a difference |
| Relative Difference  | Measuring proportional changes          |
| Rate Features        | Measuring changes over time             |
| Normalization        | Bringing values to a common scale       |

## Conclusion

Numerical Feature Engineering helps convert raw numerical columns into meaningful business measurements.

The correct technique depends on the data distribution, business objective and Machine Learning algorithm.


```python
import pandas as pd
import numpy as np

# 1. Create Sample Telecom Dataset

data = {
    "CustomerID": [101, 102, 103, 104, 105],
    "MonthlyCharges": [50, 80, 100, 40, 90],
    "Tenure": [12, 24, 36, 6, 48],
    "TotalCharges": [600, 1920, 3600, 240, 4320],
    "InternetCharges": [30, 50, 70, 20, 60],
    "PhoneCharges": [20, 30, 30, 20, 30],
    "PreviousCharges": [45, 75, 90, 35, 85],
    "PreviousMonthPurchases": [4, 6, 8, 2, 7],
    "CurrentMonthPurchases": [5, 8, 10, 3, 9]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)


# 2. Addition

df["Combined_Service_Charges"] = (
    df["InternetCharges"] + df["PhoneCharges"]
)


# 3. Subtraction

df["Service_Cost_Difference"] = (
    df["InternetCharges"] - df["PhoneCharges"]
)


# 4. Multiplication

df["Estimated_Total_Spending"] = (
    df["MonthlyCharges"] * df["Tenure"]
)


# 5. Division

df["Average_Monthly_Spending"] = (
    df["TotalCharges"] / df["Tenure"].replace(0, np.nan)
)


# 6. Ratio

df["Internet_Phone_Ratio"] = (
    df["InternetCharges"] /
    df["PhoneCharges"].replace(0, np.nan)
)


# 7. Percentage

df["Internet_Charge_Percentage"] = (
    df["InternetCharges"] /
    df["MonthlyCharges"].replace(0, np.nan)
) * 100


# 8. Difference

df["Monthly_Charge_Change"] = (
    df["MonthlyCharges"] - df["PreviousCharges"]
)


# 9. Aggregations

df["Total_Service_Charges"] = df[
    ["InternetCharges", "PhoneCharges"]
].sum(axis=1)

df["Average_Service_Charges"] = df[
    ["InternetCharges", "PhoneCharges"]
].mean(axis=1)

df["Maximum_Service_Charge"] = df[
    ["InternetCharges", "PhoneCharges"]
].max(axis=1)


# 10. Log Transformation

df["Log_Total_Charges"] = np.log1p(df["TotalCharges"])


# 11. Square Root Transformation

df["Sqrt_Total_Charges"] = np.sqrt(df["TotalCharges"])


# 12. Power Transformation

df["Squared_Monthly_Charges"] = (
    df["MonthlyCharges"] ** 2
)


# 13. Absolute Difference

df["Absolute_Charge_Difference"] = np.abs(
    df["MonthlyCharges"] - df["PreviousCharges"]
)


# 14. Relative Difference

df["Relative_Charge_Difference"] = (
    df["MonthlyCharges"] - df["PreviousCharges"]
) / df["PreviousCharges"].replace(0, np.nan)


# 15. Rate Feature

df["Monthly_Purchase_Rate"] = (
    df["CurrentMonthPurchases"] / 1
)


# 16. Normalization

minimum = df["MonthlyCharges"].min()
maximum = df["MonthlyCharges"].max()

df["Normalized_Monthly_Charges"] = (
    (df["MonthlyCharges"] - minimum) /
    (maximum - minimum)
)


# Display Final Dataset

print("\nEngineered Dataset:")
display(df)

print("\nOriginal Dataset Shape:", (5, 9))
print("Engineered Dataset Shape:", df.shape)

print("\nNewly Created Features:")
print(df.columns[9:].tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

## 4.2 Leakage Analysis

Before using these features in Machine Learning, we must check whether the source information is available at the time of prediction.

For example:

* Monthly charges should be available at prediction time.
* Tenure should represent tenure as of the prediction date.
* Previous charges should come from an earlier period.
* Current charges should not contain information recorded after the prediction date.

We must also avoid calculating customer-level statistics using future transactions.
