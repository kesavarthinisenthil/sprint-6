# Text Feature Engineering

## 1. What is Text Feature Engineering?

Text Feature Engineering is the process of converting raw text data into meaningful numerical features that can be used by Machine Learning algorithms.

Machine Learning models generally cannot directly understand text sentences.

Therefore, we convert text into numerical representations.

Example:

Customer Feedback:

"The internet connection is very slow."

After text processing, we can extract:

- Number of words
- Number of characters
- Number of positive words
- Number of negative words
- TF-IDF values

These features help Machine Learning models understand text patterns.

## 2. Why is Text Feature Engineering Important?

Text data is available in many business applications:

- Customer reviews
- Emails
- Chat messages
- Social media comments
- Customer complaints
- Product descriptions
- Support tickets

Text Feature Engineering helps:

- Understand customer opinions.
- Identify complaints.
- Analyze customer satisfaction.
- Classify messages.
- Detect spam.
- Support sentiment analysis.

## 3. Text Cleaning

Text cleaning removes unnecessary information from raw text.

Common text-cleaning operations:

- Convert text into lowercase.
- Remove punctuation.
- Remove special characters.
- Remove extra spaces.
- Remove unnecessary numbers when appropriate.
- Handle missing text values.

Example:

Original:

"THE Internet Connection is VERY Slow!!!"

After cleaning:

"the internet connection is very slow"

## 4. Tokenization

Tokenization is the process of splitting text into smaller units called tokens.

Example:

Sentence:

"Machine learning is useful"

Tokens:

["Machine", "learning", "is", "useful"]

Tokens can be words, characters, or subwords.

## 5. Stopword Removal

Stopwords are commonly occurring words that may provide limited information for certain text-analysis tasks.

Examples:

- is
- the
- a
- an
- and
- of

Example:

Original:

"The internet connection is very slow"

After stopword removal:

["internet", "connection", "slow"]

Important: Stopwords should not always be removed. Words such as "not" and "no" can change the meaning of a sentence.

## 6. Stemming

Stemming reduces words to their root-like forms by removing word endings.

Examples:

- Playing → Play
- Played → Play
- Connected → Connect

Stemming is useful when different forms of a word should be treated similarly.

However, stemming can sometimes produce words that are not grammatically correct.

## 7. Lemmatization

Lemmatization converts words into their meaningful dictionary base forms.

Examples:

- Running → Run
- Better → Good
- Mice → Mouse

Lemmatization generally produces more meaningful results than simple stemming but may require additional language resources.

## 8. Bag of Words (BoW)

Bag of Words represents text based on the frequency of words.

Example:

Documents:

1. "Good internet service"
2. "Bad internet service"

Vocabulary:

["good", "internet", "service", "bad"]

Numerical representation:

| Document | good | internet | service | bad |
|---|---:|---:|---:|---:|
| Document 1 | 1 | 1 | 1 | 0 |
| Document 2 | 0 | 1 | 1 | 1 |

Bag of Words focuses on word frequency and does not preserve word order.

## 9. TF-IDF

TF-IDF stands for Term Frequency–Inverse Document Frequency.

It measures how important a word is in a document compared with a collection of documents.

TF-IDF gives higher importance to words that are frequent in a particular document but less common across the entire collection.

It is commonly used in:

- Text classification
- Search engines
- Document analysis
- Spam detection
- Customer feedback analysis

## 10. N-Grams

N-grams are groups of consecutive words.

Types:

- Unigram: One word
- Bigram: Two consecutive words
- Trigram: Three consecutive words

Example:

Sentence:

"Customer service is good"

Unigrams:

["Customer", "service", "is", "good"]

Bigrams:

["Customer service", "service is", "is good"]

Trigrams:

["Customer service is", "service is good"]

N-grams help capture simple word relationships.

## 11. Text-Based Numerical Features

We can create numerical features from text, such as:

- Word count
- Character count
- Average word length
- Number of uppercase letters
- Number of exclamation marks
- Number of question marks
- Positive keyword count
- Negative keyword count

These features can help identify patterns in customer communication.

## 12. Important Considerations

- Clean text carefully.
- Preserve important words such as "not" and "never".
- Avoid removing useful business information.
- Fit text vectorizers using training data only.
- Apply the same transformation to test data.
- Avoid using future customer messages when predicting an earlier outcome.
- Evaluate whether each text feature improves the model.

## Conclusion

Text Feature Engineering converts unstructured text into numerical information.

Techniques such as cleaning, tokenization, Bag of Words, TF-IDF, and N-grams help Machine Learning models work with text data.

# Demonstration: Customer Feedback Analysis

## Business Scenario

A telecom company receives customer feedback through support messages and reviews.

The company wants to understand whether customers are satisfied or dissatisfied.

Example customer feedback:

1. "The internet speed is excellent."
2. "My connection is very slow."
3. "Customer support was helpful."
4. "I am unhappy with the service."
5. "The network is not working properly."

Raw text cannot be directly processed by most traditional Machine Learning algorithms.

We need to convert the text into useful numerical features.

## Example 1: Text Cleaning

Original:

"THE Internet Connection is VERY Slow!!!"

Cleaned:

"the internet connection is very slow"

Feature Name: `Cleaned_Feedback`

Reason: Removes unnecessary text variations and punctuation.

## Example 2: Word Count

Text:

"The internet connection is slow"

Word Count:

5

Feature Name: `Word_Count`

Reason: Helps understand the length of customer feedback.

## Example 3: Character Count

Text:

"Internet is slow"

Character Count:

17 characters, including spaces.

Feature Name: `Character_Count`

Reason: Helps measure the size of a text message.

## Example 4: Keyword Count

Text:

"The service is very slow and the support is poor"

Negative keywords:

- slow
- poor

Negative Keyword Count = 2

Feature Name: `Negative_Word_Count`

Reason: Helps identify potentially dissatisfied customers.

## Example 5: Bag of Words

Documents:

- "Good internet service"
- "Bad internet service"

Vocabulary:

good, internet, service, bad

Each document is represented using word counts.

## Example 6: TF-IDF

TF-IDF gives importance to words based on their frequency within a document and their rarity across the document collection.

It can help identify important terms in customer complaints.

## Example 7: N-Grams

Sentence:

"Internet connection is slow"

Bigrams:

- Internet connection
- connection is
- is slow

Bigrams help capture simple relationships between consecutive words.

## Business Value

Text Feature Engineering helps businesses:

- Analyze customer feedback.
- Identify common complaints.
- Understand customer satisfaction.
- Categorize support messages.
- Identify potential service problems.

## ML Value

Text features can be used for:

- Sentiment analysis.
- Text classification.
- Spam detection.
- Customer complaint classification.
- Customer churn prediction.

Important: Keyword counts are simple indicators, not reliable sentiment predictions by themselves. Their usefulness must be validated.

```python
import pandas as pd
import numpy as np
import re

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# ---------------------------------------
# 1. Create sample customer feedback data
# ---------------------------------------

data = {
    "CustomerID": [101, 102, 103, 104, 105, 106],

    "Feedback": [
        "The internet connection is very slow!",
        "Excellent internet service and good speed.",
        "I am unhappy with the poor customer support.",
        "The network is working perfectly.",
        "My internet is not working properly.",
        "Customer service was very helpful."
    ]
}

df = pd.DataFrame(data)

print("Original Customer Feedback:")
print(df)


# ---------------------------------------
# 2. Text Cleaning
# ---------------------------------------

def clean_text(text):

    text = text.lower()

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z\s]", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


df["Cleaned_Feedback"] = df["Feedback"].apply(clean_text)

print("\nCleaned Feedback:")
print(df[["Feedback", "Cleaned_Feedback"]])


# ---------------------------------------
# 3. Tokenization
# ---------------------------------------

df["Tokens"] = df["Cleaned_Feedback"].apply(
    lambda text: text.split()
)

print("\nTokenized Feedback:")
print(df[["Cleaned_Feedback", "Tokens"]])


# ---------------------------------------
# 4. Stopword Removal
# ---------------------------------------

stopwords = {
    "the", "is", "a", "an", "and",
    "was", "am", "with", "my"
}

df["Filtered_Tokens"] = df["Tokens"].apply(
    lambda tokens: [
        word for word in tokens
        if word not in stopwords
    ]
)

print("\nAfter Stopword Removal:")
print(df[["Tokens", "Filtered_Tokens"]])


# ---------------------------------------
# 5. Word Count
# ---------------------------------------

df["Word_Count"] = df["Tokens"].apply(len)

print("\nWord Count:")
print(df[["CustomerID", "Word_Count"]])


# ---------------------------------------
# 6. Character Count
# ---------------------------------------

df["Character_Count"] = df["Cleaned_Feedback"].apply(len)

print("\nCharacter Count:")
print(df[["CustomerID", "Character_Count"]])


# ---------------------------------------
# 7. Average Word Length
# ---------------------------------------

df["Average_Word_Length"] = df["Tokens"].apply(
    lambda tokens: (
        np.mean([len(word) for word in tokens])
        if len(tokens) > 0 else 0
    )
)

print("\nAverage Word Length:")
print(df[["CustomerID", "Average_Word_Length"]])


# ---------------------------------------
# 8. Positive and Negative Keyword Counts
# ---------------------------------------

positive_words = {
    "good", "excellent", "perfectly",
    "helpful", "great", "happy"
}

negative_words = {
    "slow", "poor", "unhappy",
    "bad", "problem", "issue"
}

df["Positive_Word_Count"] = df["Tokens"].apply(
    lambda tokens: sum(
        word in positive_words for word in tokens
    )
)

df["Negative_Word_Count"] = df["Tokens"].apply(
    lambda tokens: sum(
        word in negative_words for word in tokens
    )
)

print("\nKeyword Counts:")
print(df[
    [
        "CustomerID",
        "Positive_Word_Count",
        "Negative_Word_Count"
    ]
])


# ---------------------------------------
# 9. Bag of Words
# ---------------------------------------

count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(
    df["Cleaned_Feedback"]
)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

print("\nBag of Words:")
print(bow_df)


# ---------------------------------------
# 10. TF-IDF
# ---------------------------------------

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(
    df["Cleaned_Feedback"]
)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF Representation:")
print(tfidf_df.round(2))


# ---------------------------------------
# 11. N-Grams
# ---------------------------------------

bigram_vectorizer = CountVectorizer(
    ngram_range=(2, 2)
)

bigram_matrix = bigram_vectorizer.fit_transform(
    df["Cleaned_Feedback"]
)

bigram_df = pd.DataFrame(
    bigram_matrix.toarray(),
    columns=bigram_vectorizer.get_feature_names_out()
)

print("\nBigram Representation:")
print(bigram_df)


# ---------------------------------------
# 12. Final Feature Dataset
# ---------------------------------------

final_features = df[
    [
        "CustomerID",
        "Cleaned_Feedback",
        "Word_Count",
        "Character_Count",
        "Average_Word_Length",
        "Positive_Word_Count",
        "Negative_Word_Count"
    ]
].copy()

print("\nFinal Text Feature Dataset:")
print(final_features)

print("\nFinal Dataset Shape:")
print(final_features.shape)

print("\nMissing Values:")
print(final_features.isnull().sum())
```

## 2. Important Observations

- Tokenization separates sentences into individual words.
- Stopword removal can reduce unnecessary text.
- Word count and character count create simple numerical features.
- Bag of Words represents word frequency.
- TF-IDF measures the relative importance of words.
- N-grams capture combinations of consecutive words.

## 3. Feature Selection Decisions

### Features to Retain

- Cleaned_Feedback
- Word_Count
- Character_Count
- Bag of Words
- TF-IDF

These features provide useful information for text analysis and classification.

### Features Requiring Further Analysis

- Average_Word_Length
- Positive_Word_Count
- Negative_Word_Count
- Bigrams

Their usefulness should be evaluated using validation results and model performance.

## 4. Leakage Prevention

Text vectorizers must be fitted only on training data.

Correct process:

1. Split the dataset into training and testing sets.
2. Fit the text vectorizer on training text.
3. Transform training text.
4. Transform testing text using the same fitted vectorizer.
5. Evaluate the model.

Do not fit TF-IDF or Bag of Words on the complete dataset before splitting.

Also, customer feedback must have been available at the time the prediction is made. Feedback submitted after a churn event must not be used to predict that earlier churn event.
