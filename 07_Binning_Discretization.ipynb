# Binning & Discretization

## 1. What is Binning?

Binning is a feature engineering technique used to convert continuous numerical values into a smaller number of categories or groups.

For example, customer ages can be divided into groups:

- 18–25: Young
- 26–35: Adult
- 36–50: Middle-aged
- Above 50: Senior

Instead of using the exact age, we can use the age category.

## 2. What is Discretization?

Discretization is the process of converting continuous numerical data into discrete intervals or categories.

Example:

Original values:

22, 25, 28, 35, 42, 55

After discretization:

Young, Young, Adult, Adult, Middle-aged, Senior

Binning is a common method of discretization.

## 3. Why is Binning Important?

Binning can help:

- Simplify numerical data.
- Reduce the effect of small variations.
- Identify meaningful customer groups.
- Improve interpretability.
- Represent business categories.
- Discover patterns between groups.

However, binning can also remove useful information if the boundaries are poorly selected.

## 4. Equal-Width Binning

Equal-width binning divides the numerical range into intervals of equal size.

Formula:

Bin Width = (Maximum Value - Minimum Value) / Number of Bins

Example:

Age values range from 20 to 60.

Number of bins = 4

Bin width = (60 - 20) / 4 = 10

The bins are:

- 20–30
- 30–40
- 40–50
- 50–60

In Pandas, we can use `pd.cut()`.

## 5. Equal-Frequency Binning

Equal-frequency binning divides the data into groups containing approximately the same number of observations.

Example:

If we have 100 customers and create 4 bins, each bin will contain approximately 25 customers.

In Pandas, we can use:

`pd.qcut()`

This method uses quantiles to create the bins.

## 6. Custom Binning

Custom binning allows us to define meaningful category boundaries based on business requirements.

Example:

Monthly Charges:

- Below ₹300: Low
- ₹300–₹500: Medium
- Above ₹500: High

Custom binning is useful when business knowledge determines the category boundaries.

## 7. pd.cut()

`pd.cut()` divides numerical values into specified intervals.

It is commonly used for equal-width and custom binning.

Example:

`pd.cut(data, bins=3)`

## 8. pd.qcut()

`pd.qcut()` divides numerical data into approximately equal-sized groups based on ranks or quantiles.

Example:

`pd.qcut(data, q=4)`

This creates four quantile-based groups.

## 9. Numerical to Categorical Conversion

Binning changes numerical values into categories.

Example:

Original feature:

Customer Tenure = 45 months

After binning:

Tenure Group = Long-Term Customer

This makes the feature easier to interpret.

## 10. Advantages of Binning

- Simplifies numerical information.
- Makes data easier to understand.
- Helps identify customer segments.
- Can reduce sensitivity to small numerical changes.
- Supports business interpretation.

## 11. Disadvantages of Binning

- Loses detailed numerical information.
- Poor boundaries can create misleading categories.
- Different binning methods can produce different results.
- May reduce model performance.
- Can hide useful patterns.

## 12. Important Considerations

- Select bin boundaries carefully.
- Use business knowledge when appropriate.
- Check the number of observations in each bin.
- Avoid creating too many categories.
- Fit data-driven bin boundaries using training data only.
- Apply the same boundaries to testing and production data.
- Compare model performance before and after binning.

## Conclusion

Binning and Discretization convert continuous numerical features into meaningful groups.

Methods such as equal-width, equal-frequency, and custom binning help simplify data and identify customer segments.

However, binning should only be retained when it provides useful business or Machine Learning value.

# Demonstration: Binning in a Telecom Business

## Business Scenario

A telecom company wants to divide customers into different groups based on their age, monthly charges, and tenure.

Instead of analyzing only numerical values, the company wants meaningful customer categories.

## Example 1: Age Binning

| Age | Age Group |
|---:|---|
| 22 | Young |
| 28 | Adult |
| 35 | Adult |
| 42 | Middle-aged |
| 55 | Senior |

Feature Name: `Age_Group`

Reason: Helps identify customer age segments.

## Example 2: Monthly Charges Binning

| Monthly Charges | Category |
|---:|---|
| 200 | Low |
| 350 | Medium |
| 450 | Medium |
| 600 | High |
| 800 | High |

Feature Name: `Charge_Category`

Reason: Helps identify low-, medium-, and high-spending customers.

## Example 3: Equal-Width Binning

Suppose customer tenure values range from 0 to 60 months.

We divide them into three equal-width intervals:

- 0–20 months: Short-Term
- 20–40 months: Medium-Term
- 40–60 months: Long-Term

Feature Name: `Tenure_EqualWidth`

Reason: Divides customers based on tenure duration.

## Example 4: Equal-Frequency Binning

Suppose we have 12 customers.

We divide them into three groups with approximately four customers in each group.

- Low Tenure
- Medium Tenure
- High Tenure

Feature Name: `Tenure_EqualFrequency`

Reason: Creates groups based on the distribution of customer tenure.

## Example 5: Custom Binning

A business may decide that customers paying below ₹300 belong to the low-charge category.

The boundaries are selected using business knowledge rather than automatically calculated intervals.

## Business Value

Binning helps telecom companies:

- Identify customer segments.
- Understand spending groups.
- Categorize customer tenure.
- Simplify customer analysis.
- Support business decisions.

## ML Value

Categorical groups may help some Machine Learning models identify patterns more easily.

However, numerical information can be lost during binning. Therefore, the original numerical feature should usually be retained until model evaluation confirms whether the binned feature is useful.

```python
import pandas as pd
import numpy as np

# ---------------------------------------
# 1. Create sample customer dataset
# ---------------------------------------

data = {
    "CustomerID": [101, 102, 103, 104, 105, 106,
                   107, 108, 109, 110, 111, 112],

    "Age": [22, 25, 28, 32, 35, 38,
            42, 45, 48, 52, 56, 60],

    "MonthlyCharges": [
        200, 250, 300, 350, 400, 450,
        500, 550, 600, 650, 700, 800
    ],

    "Tenure": [2, 5, 8, 12, 15, 20,
               25, 30, 35, 42, 50, 60]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# ---------------------------------------
# 2. Equal-width binning using pd.cut()
# ---------------------------------------

df["Age_EqualWidth"] = pd.cut(
    df["Age"],
    bins=4,
    labels=[
        "Young",
        "Adult",
        "Middle_Aged",
        "Senior"
    ],
    include_lowest=True
)

# ---------------------------------------
# 3. Equal-frequency binning using pd.qcut()
# ---------------------------------------

df["Tenure_EqualFrequency"] = pd.qcut(
    df["Tenure"],
    q=3,
    labels=[
        "Low_Tenure",
        "Medium_Tenure",
        "High_Tenure"
    ]
)

# ---------------------------------------
# 4. Custom binning for monthly charges
# ---------------------------------------

charge_bins = [0, 300, 500, 1000]

charge_labels = [
    "Low_Charges",
    "Medium_Charges",
    "High_Charges"
]

df["Charge_Category"] = pd.cut(
    df["MonthlyCharges"],
    bins=charge_bins,
    labels=charge_labels,
    include_lowest=True
)

# ---------------------------------------
# 5. Custom binning for customer tenure
# ---------------------------------------

tenure_bins = [0, 12, 36, 100]

tenure_labels = [
    "New_Customer",
    "Regular_Customer",
    "Long_Term_Customer"
]

df["Tenure_Category"] = pd.cut(
    df["Tenure"],
    bins=tenure_bins,
    labels=tenure_labels,
    include_lowest=True
)

# ---------------------------------------
# 6. Create a binary feature
# ---------------------------------------

df["High_Value_Customer"] = (
    df["MonthlyCharges"] >= 500
).astype(int)

# ---------------------------------------
# 7. Display engineered dataset
# ---------------------------------------

print("\nDataset After Binning:")
print(df)

print("\nDataset Shape:")
print(df.shape)

# ---------------------------------------
# 8. Check category distributions
# ---------------------------------------

print("\nAge Group Distribution:")
print(df["Age_EqualWidth"].value_counts())

print("\nTenure Group Distribution:")
print(df["Tenure_EqualFrequency"].value_counts())

print("\nCharge Category Distribution:")
print(df["Charge_Category"].value_counts())

print("\nTenure Category Distribution:")
print(df["Tenure_Category"].value_counts())

# ---------------------------------------
# 9. Check missing values
# ---------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# ---------------------------------------
# 10. Display only new features
# ---------------------------------------

new_features = [
    "Age_EqualWidth",
    "Tenure_EqualFrequency",
    "Charge_Category",
    "Tenure_Category",
    "High_Value_Customer"
]

print("\nNewly Created Binning Features:")
print(df[["CustomerID"] + new_features])
```

## 4. Leakage Prevention

Binning boundaries must be calculated using only the training dataset when they are learned from data distributions.

For example:

If we use `pd.qcut()` on the entire dataset before splitting it into training and testing data, information from the testing distribution can influence the bin boundaries.

A safer process is:

1. Split the dataset into training and testing sets.
2. Learn bin boundaries using training data.
3. Apply the same boundaries to testing data.
4. Evaluate the model.

Business-defined boundaries can be used when they are established independently of the test data.
