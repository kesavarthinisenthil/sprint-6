# 1. Introduction to Feature Engineering

## 1.1 What is a Feature?

A feature is an individual piece of information or an attribute that a Machine Learning model uses to understand patterns and make predictions.

For example, in a customer dataset, Age, Monthly Charges, Tenure, and Contract Type are features.

## 1.2 Feature vs Variable

* **Feature:** An input column specifically used by a Machine Learning model.
* **Variable:** Any value or attribute that can change in a dataset.

Every feature is a variable, but not every variable must be used as a model feature.

## 1.3 Feature vs Target

* **Feature:** Input information given to the model to make predictions.
* **Target:** The output that the model needs to predict.

Example:

In a customer churn prediction problem:

* Age – Feature
* Tenure – Feature
* Monthly Charges – Feature
* Churn – Target

The model uses customer information to predict whether a customer will leave the company.

## 1.4 What is Feature Engineering?

Feature Engineering is the process of creating, transforming, and improving input features from existing raw data to help Machine Learning models learn better patterns.

It involves using mathematical operations, business knowledge, and data understanding to generate meaningful information.

## 1.5 Why Feature Engineering?

Feature Engineering is important because:

1. It helps models identify useful patterns.
2. It improves prediction performance.
3. It converts raw information into meaningful features.
4. It reduces unnecessary information.
5. It helps models understand business-related patterns.
6. It can make simple Machine Learning algorithms perform better.

## 1.6 Importance of Domain Knowledge

Domain knowledge means understanding the industry or business area related to the dataset.

It helps us create features that have real-world meaning.

Example:

In telecom, customer tenure and monthly charges can help understand customer churn.

A feature such as Total Customer Spending can be created using:

Total Spending = Monthly Charges × Tenure

This feature may help identify customers who have spent more money over time.

## 1.7 Good Features vs Bad Features

### Good Features

* Provide useful information.
* Have a meaningful relationship with the target.
* Represent real-world information.
* Do not contain data leakage.
* Help the model identify patterns.

### Bad Features

* Contain irrelevant information.
* Have many missing or incorrect values.
* Duplicate information unnecessarily.
* Introduce data leakage.
* Add unnecessary complexity.

## 1.8 Feature Engineering Workflow

The Feature Engineering workflow includes:

1. Understand the dataset.
2. Identify existing features.
3. Understand the target variable.
4. Identify missing information.
5. Create meaningful features.
6. Transform existing features.
7. Check feature quality.
8. Perform feature selection.
9. Check for feature leakage.
10. Prepare the final ML-ready dataset.

## 1.9 Feature Creation

Feature Creation means generating new features using existing columns.

Example:

Total Spending = Monthly Charges × Tenure

This creates a new column that represents estimated cumulative customer charges.

## 1.10 Feature Transformation

Feature Transformation means changing the format or distribution of existing features.

Examples:

* Log transformation
* Standardization
* Normalization
* Square root transformation

These transformations can help Machine Learning algorithms work better with different types of data.

## 1.11 Feature Selection

Feature Selection is the process of choosing the most useful features for a Machine Learning model and removing irrelevant or redundant features.

Example:

If a dataset contains Customer ID, Age, Tenure, Monthly Charges, and Churn, Customer ID may not provide meaningful predictive information and may be excluded.

## 1.12 Feature Extraction

Feature Extraction means converting existing information into a new set of meaningful features.

Example:

A date column such as `2026-11-04` can be converted into:

* Year: 2026
* Month: 11
* Day: 4
* Day of Week: Wednesday

## 1.13 Feature Leakage

Feature Leakage happens when a Machine Learning model receives information during training that would not actually be available when making real-world predictions.

Example:

If we want to predict whether a customer will leave, using a column that records the customer's cancellation date could reveal the answer directly.

This creates misleading model performance.

## 1.14 Role of Feature Engineering in the ML Pipeline

The Machine Learning pipeline generally follows:

Raw Data → Data Cleaning → Exploratory Data Analysis → Feature Engineering → Feature Selection → Model Training → Model Evaluation → Prediction

Feature Engineering converts existing data into useful information that helps Machine Learning models make better predictions.

## Conclusion

Feature Engineering is an important part of Machine Learning because the quality and usefulness of input features influence how well a model can learn.

An AI/ML Engineer should understand not only how to create new features but also why they are created, what information they represent, and whether they introduce any data leakage.


# Demonstration of Feature Engineering

However, creating more features does not always guarantee better performance. Each feature must be evaluated.

```python
import pandas as pd
import numpy as np

# Create sample customer dataset
data = {
    "CustomerID": [101, 102, 103, 104, 105],
    "Tenure": [12, 24, 36, 6, 48],
    "MonthlyCharges": [50, 80, 100, 40, 90],
    "TotalCharges": [600, 1920, 3600, 240, 4320],
    "Contract": ["Month-to-month", "One year", "Two year",
                 "Month-to-month", "Two year"],
    "Churn": ["Yes", "No", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

# Feature Creation
df["Estimated_Total_Spending"] = df["MonthlyCharges"] * df["Tenure"]

# Feature Transformation
df["Tenure_Group"] = pd.cut(
    df["Tenure"],
    bins=[0, 12, 36, 100],
    labels=["New", "Medium", "Long-Term"],
    include_lowest=True
)

# Feature Creation using condition
df["Long_Term_Customer"] = np.where(
    df["Tenure"] >= 24, 1, 0
)

# Feature Creation using division
df["Average_Monthly_Spending"] = (
    df["TotalCharges"] / df["Tenure"]
)

print("\nEngineered Dataset:")
display(df)

print("\nDataset Shape:", df.shape)

print("\nNewly Created Features:")
print([
    "Estimated_Total_Spending",
    "Tenure_Group",
    "Long_Term_Customer",
    "Average_Monthly_Spending"
])
```

## 4.1 Feature Documentation

### Feature 1: Estimated_Total_Spending

* **Reason:** To estimate the total amount spent by a customer.
* **Business Meaning:** Represents the estimated cumulative customer spending.
* **ML Relevance:** May help identify spending patterns associated with customer churn.
* **Leakage Risk:** Low, provided both source columns are available at prediction time.
* **Final Decision:** Retain.

### Feature 2: Tenure_Group

* **Source Columns:** Tenure
* **Logic:** Group customers into New, Medium and Long-Term categories.
* **Reason:** To identify different customer segments.
* **Business Meaning:** Represents the length of the customer's relationship with the company.
* **ML Relevance:** May help identify differences in churn behaviour across customer groups.
* **Leakage Risk:** Low.
* **Final Decision:** Retain.

### Feature 3: Long_Term_Customer

* **Source Columns:** Tenure
* **Logic:** If Tenure >= 24, assign 1; otherwise, assign 0.
* **Reason:** To identify customers who have stayed for at least 24 months.
* **Business Meaning:** Represents established customers.
* **ML Relevance:** May help the model identify relationships between customer loyalty and churn.
* **Leakage Risk:** Low.
* **Final Decision:** Retain.

### Feature 4: Average_Monthly_Spending

* **Source Columns:** TotalCharges, Tenure
* **Logic:** TotalCharges / Tenure
* **Reason:** To calculate the average monthly spending of a customer.
* **Business Meaning:** Represents the average amount charged per month.
* **ML Relevance:** May help identify customer spending patterns.
* **Leakage Risk:** Low, provided the information is available at prediction time.
* **Final Decision:** Retain.
