# Date & Time Feature Engineering

## 1. What is Date & Time Feature Engineering?

Date and Time Feature Engineering is the process of extracting useful information from date and time columns and converting it into meaningful numerical or categorical features.

For example, instead of using only a customer's signup date, we can extract:

* Signup year
* Signup month
* Signup day
* Day of the week
* Weekend or weekday
* Customer tenure
* Days since the last purchase

These features can help Machine Learning models understand customer behaviour and time-based patterns.

## 2. Why is Date & Time Feature Engineering Important?

Raw date values are not always directly useful to Machine Learning algorithms.

For example:

`2024-05-15`

We can extract:

* Year: 2024
* Month: 5
* Day: 15
* Day of Week: Wednesday
* Quarter: Q2

These extracted values may help identify seasonal patterns, customer activity, and business trends.

## 3. Year, Month and Day

We can extract the year, month, and day from a date column.

Example:

Date: `2025-08-20`

* Year = 2025
* Month = 8
* Day = 20

**ML Use:** Helps identify yearly, monthly, or daily patterns.

## 4. Day of the Week

The day of the week represents whether an event happened on Monday, Tuesday, Wednesday, and so on.

Python uses:

* Monday = 0
* Tuesday = 1
* Wednesday = 2
* Thursday = 3
* Friday = 4
* Saturday = 5
* Sunday = 6

**ML Use:** Useful for understanding customer activity patterns.

## 5. Week Number and Quarter

A year contains 52 or 53 calendar weeks and four quarters.

* Q1: January to March
* Q2: April to June
* Q3: July to September
* Q4: October to December

**ML Use:** Helps identify quarterly and seasonal business patterns.

## 6. Weekend Indicator

A weekend indicator identifies whether a date falls on Saturday or Sunday.

* 0 = Weekday
* 1 = Weekend

**ML Use:** Helps identify differences between weekday and weekend customer behaviour.

## 7. Month Start and Month End

These features identify whether a date is the first or last day of a month.

Examples:

* 2026-11-01: Month Start
* 2026-11-30: Month End

**ML Use:** Useful for identifying billing cycles, monthly reporting, and payment patterns.

## 8. Days Since Event

This feature calculates the number of days between an event date and a fixed reference date.

Formula:

Days Since Event = Reference Date - Event Date

**ML Use:** Helps measure how recently a customer performed an activity.

## 9. Time Difference

Time difference calculates the duration between two dates.

Example:

Signup Date = 2024-01-01
Reference Date = 2024-12-31

Time Difference = 365 days

**ML Use:** Helps calculate customer duration and activity periods.

## 10. Customer Age

Customer age can be calculated using the date of birth and a reference date.

**ML Use:** Can help understand customer demographics when this information is available and appropriate to use.

## 11. Customer Tenure

Customer tenure represents how long a customer has been associated with a company.

Formula:

Customer Tenure = Reference Date - Signup Date

**ML Use:** Helps identify new customers and long-term customers.

## 12. Recency

Recency measures how recently a customer performed an activity.

Example:

If a customer last purchased something 10 days ago:

Recency = 10 days

**ML Use:** Helps understand customer engagement and identify inactive customers.

## 13. Frequency

Frequency represents how many times a customer performs an activity during a defined period.

Example:

A customer makes 5 purchases in the last 12 months.

Frequency = 5

**ML Use:** Helps identify frequent and occasional customers.

## 14. Seasonality

Seasonality refers to repeating patterns that occur during particular periods.

Examples:

* Monthly sales patterns
* Quarterly customer activity
* Seasonal product demand

**ML Use:** Helps models understand time-related business patterns.

## 15. Important Considerations

* Convert date columns into proper datetime format.
* Handle missing or invalid dates.
* Use a consistent reference date.
* Do not use information that would not have been available at prediction time.
* Avoid future data leakage.
* Create date features only when the source dates are reliable.

## Conclusion

Date & Time Feature Engineering converts raw dates into meaningful features such as year, month, day, tenure, recency, and frequency.

These features help Machine Learning models identify time-based patterns and improve prediction quality.


# Demonstration: Date & Time Feature Engineering in a Telecom Business

## Business Scenario

A telecom company wants to understand customer behaviour and predict customer churn.

The company has customer information such as:

* Customer signup date
* Date of birth
* Last purchase date
* Transaction dates

Instead of using only the original dates, we can extract meaningful features.

## Example 1: Extracting Date Components

Signup Date: `2024-05-15`

| Feature      |     Value |
| ------------ | --------: |
| Signup Year  |      2024 |
| Signup Month |         5 |
| Signup Day   |        15 |
| Day of Week  | Wednesday |
| Quarter      |        Q2 |

**Reason:** These features help identify when customers joined the company.

## Example 2: Customer Tenure

Signup Date: `2024-01-01`
Reference Date: `2025-01-01`

Customer Tenure = 366 days

**Reason:** Helps identify how long a customer has been associated with the company.

## Example 3: Recency

Last Purchase Date: `2026-10-20`
Reference Date: `2026-11-01`

Recency = 12 days

**Reason:** Helps identify how recently the customer purchased a service.

## Example 4: Customer Frequency

Suppose a customer makes 4 purchases within the last 12 months.

Frequency = 4

**Reason:** Helps identify frequently active customers.

## Example 5: Weekend Indicator

Signup Date: `2026-11-01`

This date falls on a Sunday.

Weekend Indicator = 1

**Reason:** Helps identify whether customers tend to sign up during weekdays or weekends.

## Important Dataset Note

The date information in this demonstration is synthetic sample data created for learning.

The Telco Customer Churn dataset used in earlier sprints contains a `tenure` column, but it does not provide customer birth dates or detailed purchase timestamps.

Do not assume that these additional date columns exist in the original dataset. Use real date columns only when they are available and reliable.

## Business Value

Date and time features can help a telecom company:

* Identify new and long-term customers.
* Understand customer activity patterns.
* Measure how recently customers interacted with the company.
* Identify frequently active customers.
* Understand monthly and quarterly trends.

## ML Value

These features may help Machine Learning models understand customer behaviour and time-related patterns.

However, every feature must be checked for availability at prediction time to prevent data leakage.


```python
import pandas as pd
import numpy as np

# ---------------------------------------
# 1. Create a sample customer dataset
# ---------------------------------------

data = {
    "CustomerID": [101, 102, 103, 104, 105],

    "SignupDate": [
        "2024-01-15",
        "2023-06-20",
        "2025-03-10",
        "2022-11-05",
        "2024-08-25"
    ],

    "BirthDate": [
        "1995-04-12",
        "1988-09-20",
        "2000-02-15",
        "1992-11-08",
        "1998-07-30"
    ],

    "LastPurchaseDate": [
        "2026-10-20",
        "2026-09-15",
        "2026-10-28",
        "2026-08-10",
        "2026-10-25"
    ]
}

df = pd.DataFrame(data)

# ---------------------------------------
# 2. Convert columns into datetime format
# ---------------------------------------

date_columns = [
    "SignupDate",
    "BirthDate",
    "LastPurchaseDate"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

# Use one fixed reference date for all customers
snapshot_date = pd.Timestamp("2026-11-01")

# ---------------------------------------
# 3. Extract year, month and day
# ---------------------------------------

df["Signup_Year"] = df["SignupDate"].dt.year

df["Signup_Month"] = df["SignupDate"].dt.month

df["Signup_Day"] = df["SignupDate"].dt.day

# ---------------------------------------
# 4. Extract day of the week
# ---------------------------------------

df["Signup_DayOfWeek"] = (
    df["SignupDate"].dt.dayofweek
)

# ---------------------------------------
# 5. Extract week number and quarter
# ---------------------------------------

df["Signup_WeekNumber"] = (
    df["SignupDate"].dt.isocalendar().week
)

df["Signup_Quarter"] = (
    df["SignupDate"].dt.quarter
)

# ---------------------------------------
# 6. Create weekend indicator
# ---------------------------------------

df["Signup_IsWeekend"] = (
    df["SignupDate"].dt.dayofweek >= 5
).astype(int)

# ---------------------------------------
# 7. Identify month start and month end
# ---------------------------------------

df["Signup_IsMonthStart"] = (
    df["SignupDate"].dt.is_month_start
).astype(int)

df["Signup_IsMonthEnd"] = (
    df["SignupDate"].dt.is_month_end
).astype(int)

# ---------------------------------------
# 8. Calculate customer tenure
# ---------------------------------------

df["Customer_Tenure_Days"] = (
    snapshot_date - df["SignupDate"]
).dt.days

# ---------------------------------------
# 9. Calculate recency
# ---------------------------------------

df["Recency_Days"] = (
    snapshot_date - df["LastPurchaseDate"]
).dt.days

# ---------------------------------------
# 10. Calculate customer age
# ---------------------------------------

df["Customer_Age"] = (
    snapshot_date.year
    - df["BirthDate"].dt.year
    - (
        (
            (snapshot_date.month < df["BirthDate"].dt.month)
            |
            (
                (snapshot_date.month == df["BirthDate"].dt.month)
                &
                (snapshot_date.day < df["BirthDate"].dt.day)
            )
        ).astype(int)
    )
)

# ---------------------------------------
# 11. Create quarter-based season feature
# ---------------------------------------

df["Signup_Season"] = (
    "Q" + df["Signup_Quarter"].astype("string")
)

# ---------------------------------------
# 12. Create sample transaction data
# ---------------------------------------

transactions = pd.DataFrame({
    "CustomerID": [
        101, 101, 101, 102, 102,
        103, 103, 103, 103, 104, 105, 105
    ],

    "TransactionDate": [
        "2026-02-10", "2026-05-15", "2026-10-20",
        "2026-03-12", "2026-09-15",
        "2026-01-05", "2026-04-10", "2026-07-20",
        "2026-10-28", "2026-08-10", "2026-06-15",
        "2026-10-25"
    ],

    "Amount": [
        500, 700, 900, 400, 600,
        300, 450, 500, 750, 800, 350, 650
    ]
})

transactions["TransactionDate"] = pd.to_datetime(
    transactions["TransactionDate"],
    errors="coerce"
)

# ---------------------------------------
# 13. Calculate transaction frequency
# ---------------------------------------

window_start = snapshot_date - pd.DateOffset(months=12)

recent_transactions = transactions[
    (transactions["TransactionDate"] <= snapshot_date)
    &
    (transactions["TransactionDate"] > window_start)
]

frequency = (
    recent_transactions
    .groupby("CustomerID")
    .size()
    .reset_index(name="Frequency_12M")
)

df = df.merge(
    frequency,
    on="CustomerID",
    how="left"
)

df["Frequency_12M"] = (
    df["Frequency_12M"]
    .fillna(0)
    .astype(int)
)

# ---------------------------------------
# 14. Display engineered features
# ---------------------------------------

print("Date and Time Feature Engineering Completed!")

print("\nEngineered Dataset:")
print(df)

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nFeature Information:")
print(df.dtypes)
```

## 4. Leakage Prevention

Date and time features must only use information available at the prediction time.

For example:

If we predict customer churn on 1 November 2026, we must not use a purchase made on 10 November 2026.

Using future information can cause data leakage and produce misleading model performance.

Always define a fixed observation or snapshot date before calculating recency, tenure, and frequency.
