# Dimensionality Reduction

## 1. What is Dimensionality Reduction?

Dimensionality Reduction is the process of reducing the number of input features in a dataset while preserving as much useful information as possible.

A dataset with many features can become complex and difficult to process.

Dimensionality Reduction helps represent the original data using fewer dimensions.

Example:

Original features:

- Age
- Tenure
- MonthlyCharges
- TotalCharges
- SupportCalls
- InternetUsage
- ContractType
- PaymentMethod

After dimensionality reduction, we may represent the information using fewer transformed features called principal components.

## 2. Why is Dimensionality Reduction Important?

It helps:

- Reduce the number of dimensions.
- Simplify complex datasets.
- Reduce computational requirements.
- Remove some redundant information.
- Support data visualization.
- Reduce noise in certain situations.
- Address problems associated with high-dimensional data.

However, dimensionality reduction may also remove useful information.

## 3. Feature Selection vs Dimensionality Reduction

| Feature Selection | Dimensionality Reduction |
|---|---|
| Selects existing features | Creates new transformed dimensions |
| Keeps original feature meanings | Transformed features may be difficult to interpret |
| Example: Retain Tenure and MonthlyCharges | Example: Create PC1 and PC2 |
| Removes unnecessary input columns | Combines information from multiple columns |

## 4. What is PCA?

PCA stands for Principal Component Analysis.

It is an unsupervised dimensionality reduction technique.

PCA transforms correlated numerical features into a smaller number of new features called Principal Components.

Principal Components:

- Are combinations of original features.
- Are uncorrelated with each other.
- Are ordered by the amount of variance they explain.

## 5. How PCA Works

PCA generally follows these steps:

1. Prepare numerical features.
2. Standardize the features when their scales differ.
3. Calculate the covariance structure.
4. Identify principal directions.
5. Calculate principal components.
6. Select the required number of components.
7. Transform the original dataset.

## 6. Explained Variance

Explained Variance indicates how much of the original data's variance is represented by a principal component.

Example:

| Component | Explained Variance |
|---|---:|
| PC1 | 50% |
| PC2 | 30% |
| PC3 | 15% |
| PC4 | 5% |

If we select the first two components:

Total explained variance = 50% + 30% = 80%.

This means the two components represent 80% of the total variance in the original standardized dataset.

## 7. Cumulative Explained Variance

Cumulative Explained Variance is the running total of explained variance across principal components.

It helps decide how many components to retain.

Example:

| Components | Cumulative Variance |
|---|---:|
| PC1 | 50% |
| PC1 + PC2 | 80% |
| PC1 + PC2 + PC3 | 95% |
| PC1 + PC2 + PC3 + PC4 | 100% |

A common approach is to select enough components to retain approximately 90–95% of the variance.

The appropriate threshold depends on the task.

## 8. Standardization Before PCA

PCA is sensitive to feature scales.

Example:

- Age: 18–70
- MonthlyCharges: 20–120
- TotalCharges: 100–8000

A feature with a larger numerical scale may dominate the PCA calculation.

Standardization converts features to a comparable scale.

Scikit-learn provides:

`StandardScaler`

## 9. Other Dimensionality Reduction Techniques

### A. t-SNE

t-SNE stands for t-distributed Stochastic Neighbor Embedding.

It is commonly used to visualize high-dimensional data in two or three dimensions.

It focuses on preserving local neighbourhood relationships.

### B. UMAP

UMAP stands for Uniform Manifold Approximation and Projection.

It is another technique for nonlinear dimensionality reduction and visualization.

### C. LDA

LDA stands for Linear Discriminant Analysis.

It is a supervised dimensionality reduction technique that uses class labels to find directions that separate classes.

Unlike PCA, LDA uses target information.

## 10. Advantages of PCA

- Reduces the number of dimensions.
- Helps remove some redundancy.
- Can improve computational efficiency.
- Supports two-dimensional visualization.
- Can reduce noise in some datasets.

## 11. Disadvantages of PCA

- Principal components may be difficult to interpret.
- Some useful information may be lost.
- Requires careful preprocessing.
- Results depend on the dataset.
- PCA does not directly use target labels.

## 12. Important Considerations

- Use numerical input features.
- Handle missing values before PCA.
- Encode categorical features appropriately.
- Standardize features when their scales differ.
- Fit the scaler and PCA using training data only.
- Apply the same fitted transformations to test data.
- Exclude target variables from unsupervised PCA.
- Evaluate model performance after dimensionality reduction.

## Conclusion

Dimensionality Reduction transforms high-dimensional data into a smaller representation.

PCA is a widely used technique that creates principal components while retaining as much variance as possible.

It is useful for visualization, reducing complexity, and preparing data for Machine Learning.

# Demonstration: PCA in Telecom Customer Data

## Business Scenario

A telecom company collects several numerical customer features.

Example:

| Customer | Age | Tenure | MonthlyCharges | TotalCharges |
|---|---:|---:|---:|---:|
| 101 | 25 | 12 | 60 | 720 |
| 102 | 35 | 24 | 80 | 1920 |
| 103 | 45 | 36 | 90 | 3240 |
| 104 | 30 | 18 | 70 | 1260 |
| 105 | 50 | 48 | 100 | 4800 |

The company wants to simplify the dataset.

## Example 1: Original Dimensions

Suppose a dataset contains six numerical features:

- Age
- Tenure
- MonthlyCharges
- TotalCharges
- SupportCalls
- InternetUsage

Original dimensions:

6 features.

## Example 2: Apply Standardization

These features have different numerical scales.

For example:

- Age ranges from 18 to 70.
- TotalCharges may range from hundreds to thousands.

Standardization converts the features into comparable scales.

## Example 3: Apply PCA

PCA combines information from the original features.

Instead of six original features, we may retain three principal components.

New dimensions:

- PC1
- PC2
- PC3

## Example 4: Explained Variance

Suppose PCA produces:

| Component | Explained Variance |
|---|---:|
| PC1 | 45% |
| PC2 | 25% |
| PC3 | 15% |
| PC4 | 10% |
| PC5 | 3% |
| PC6 | 2% |

The first three components explain:

45% + 25% + 15% = 85%.

Therefore, we can represent the original six features using three components while retaining 85% of the original variance.

## Example 5: Business Value

Dimensionality Reduction helps:

- Simplify customer datasets.
- Reduce computational complexity.
- Support customer data visualization.
- Identify major patterns in customer information.

## Example 6: ML Value

PCA may help:

- Reduce redundant dimensions.
- Improve training efficiency.
- Reduce noise.
- Prepare data for certain Machine Learning algorithms.

However, PCA does not guarantee better model performance.

## Important Note

Principal Components are combinations of original features.

For example, PC1 may contain information from Age, Tenure, MonthlyCharges, and TotalCharges.

Therefore, PC1 cannot automatically be interpreted as one specific business feature.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

# ---------------------------------------
# 1. Create sample telecom dataset
# ---------------------------------------

np.random.seed(42)

n = 200

tenure = np.random.randint(1, 73, n)

monthly_charges = np.random.uniform(20, 120, n)

total_charges = (
    tenure * monthly_charges
    + np.random.normal(0, 100, n)
)

age = np.random.randint(18, 70, n)

support_calls = np.random.poisson(2, n)

internet_usage = np.random.uniform(5, 150, n)

df = pd.DataFrame({
    "Age": age,
    "Tenure": tenure,
    "MonthlyCharges": monthly_charges,
    "TotalCharges": total_charges,
    "SupportCalls": support_calls,
    "InternetUsage": internet_usage
})

print("Original Dataset:")
print(df.head())

print("\nOriginal Dataset Shape:")
print(df.shape)


# ---------------------------------------
# 2. Check Missing Values
# ---------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# ---------------------------------------
# 3. Train-Test Split
# ---------------------------------------

X_train, X_test = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)


# ---------------------------------------
# 4. Standardize the Features
# ---------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print("\nStandardization Completed!")


# ---------------------------------------
# 5. Apply PCA
# ---------------------------------------

pca = PCA()

X_train_pca = pca.fit_transform(
    X_train_scaled
)

X_test_pca = pca.transform(
    X_test_scaled
)

print("\nPCA Transformation Completed!")


# ---------------------------------------
# 6. Explained Variance
# ---------------------------------------

explained_variance = pca.explained_variance_ratio_

variance_df = pd.DataFrame({
    "Principal_Component": [
        f"PC{i+1}" for i in range(len(explained_variance))
    ],
    "Explained_Variance": explained_variance,
    "Cumulative_Variance": np.cumsum(explained_variance)
})

print("\nExplained Variance:")
print(variance_df.round(4))


# ---------------------------------------
# 7. Select Components for 90% Variance
# ---------------------------------------

pca_90 = PCA(
    n_components=0.90
)

X_train_pca_90 = pca_90.fit_transform(
    X_train_scaled
)

X_test_pca_90 = pca_90.transform(
    X_test_scaled
)

print("\nNumber of Components Selected:")
print(pca_90.n_components_)

print("\nTraining Shape After PCA:")
print(X_train_pca_90.shape)

print("\nTesting Shape After PCA:")
print(X_test_pca_90.shape)

print(
    "\nTotal Explained Variance:",
    round(sum(pca_90.explained_variance_ratio_), 4)
)


# ---------------------------------------
# 8. Visualize Explained Variance
# ---------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(explained_variance) + 1),
    np.cumsum(explained_variance),
    marker="o"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")

plt.title("PCA - Cumulative Explained Variance")

plt.axhline(
    y=0.90,
    linestyle="--",
    label="90% Variance"
)

plt.legend()
plt.grid(True)
plt.show()


# ---------------------------------------
# 9. Visualize First Two Components
# ---------------------------------------

pca_2 = PCA(
    n_components=2
)

X_2d = pca_2.fit_transform(
    X_train_scaled
)

pca_df = pd.DataFrame(
    X_2d,
    columns=["PC1", "PC2"]
)

print("\nFirst Two Principal Components:")
print(pca_df.head())

plt.figure(figsize=(8, 6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title("PCA - Two Dimensional Representation")

plt.grid(True)
plt.show()


# ---------------------------------------
# 10. Compare Original and Reduced Dimensions
# ---------------------------------------

print("\nDimension Comparison:")

print("Original Features:", df.shape[1])

print(
    "Reduced Features:",
    pca_90.n_components_
)


# ---------------------------------------
# 11. Final PCA Dataset
# ---------------------------------------

final_pca_dataset = pd.DataFrame(
    X_train_pca_90,
    columns=[
        f"PC{i+1}"
        for i in range(pca_90.n_components_)
    ]
)

print("\nFinal PCA Dataset:")
print(final_pca_dataset.head())

print("\nFinal PCA Dataset Shape:")
print(final_pca_dataset.shape)

print("\nDimensionality Reduction Completed!")
```

# Feature Justification – Dimensionality Reduction

## 1. PCA Transformation Justification

PCA transforms original numerical features into new principal components.

| Feature Name | Source Columns | Feature Logic / Formula | Reason for Creation | Expected Relationship with Target | Business Meaning | Potential ML Value | Leakage Risk | Final Decision |
|---|---|---|---|---|---|---|---|---|
| PC1 | Age, Tenure, MonthlyCharges, TotalCharges, SupportCalls, InternetUsage | First principal component | Capture the largest amount of variance | Not directly evaluated against target because PCA is unsupervised | Major pattern in customer data | Can represent substantial information in one dimension | Low when fitted using training data only | Retain if validation supports it |
| PC2 | All original numerical features | Second principal component | Capture additional variance not represented by PC1 | Not directly evaluated against target | Second major pattern in customer data | Reduces dimensionality further | Low when fitted using training data only | Retain if validation supports it |
| PC3 and later | All original numerical features | Additional principal components | Capture remaining variance | Not directly evaluated against target | Additional data variation | May preserve useful information | Low when fitted using training data only | Evaluate using explained variance and model performance |

## 2. Important Observations

- PCA reduces the number of dimensions.
- Standardization is important when feature scales differ.
- Principal components are combinations of original features.
- Explained variance measures how much variation is retained.
- Cumulative explained variance helps select components.
- PCA does not use the target variable.

## 3. Feature Selection Decisions

### Potentially Retain

Principal components that preserve sufficient information and support good validation performance.

### Potentially Remove

Original features from the PCA-transformed model input after the transformation has been applied.

The original dataset should still be preserved for business interpretation and comparison.

### Further Analysis Required

The number of components should be selected based on:

- Explained variance.
- Model performance.
- Computational requirements.
- Interpretability needs.

## 4. Leakage Prevention

PCA must be fitted using training data only.

Correct workflow:

1. Split the dataset into training and testing data.
2. Fit StandardScaler using training data.
3. Transform training data.
4. Transform testing data using the fitted scaler.
5. Fit PCA using standardized training data.
6. Transform training and testing data using the fitted PCA model.
7. Evaluate the model.

Never fit the scaler or PCA on the complete dataset before splitting.

## 5. Final Conclusion

Dimensionality Reduction helps simplify high-dimensional datasets.

PCA combines information from multiple numerical features into a smaller number of principal components.

It can improve computational efficiency and support data visualization.

However, principal components are less interpretable than the original features.

Therefore, PCA should be used when its benefits are supported by validation results and business requirements.
