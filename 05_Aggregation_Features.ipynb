# Aggregation Features

## 1. What are Aggregation Features?

Aggregation Feature Engineering is the process of combining multiple values or records into a summarized value.

It helps convert detailed data into meaningful information.

For example, suppose a customer has three monthly bills:

* January: ₹500
* February: ₹700
* March: ₹800

We can calculate:

* Total spending = ₹2,000
* Average spending = ₹666.67
* Maximum spending = ₹800
* Minimum spending = ₹500
* Number of bills = 3

These summarized values become new features for Machine Learning.

## 2. Why are Aggregation Features Important?

Aggregation features help identify patterns that may not be visible in individual records.

They can help businesses understand:

* Total customer spending.
* Average customer spending.
* Number of transactions.
* Maximum and minimum purchases.
* Customer activity.
* Product performance.
* Monthly and yearly trends.

## 3. Common Aggregation Functions

### sum()

Calculates the total of multiple values.

Example: Total customer spending.

### mean()

Calculates the average value.

Example: Average monthly charges.

### count()

Counts the number of records.

Example: Number of transactions made by a customer.

### min()

Finds the smallest value.

Example: Lowest monthly bill.

### max()

Finds the largest value.

Example: Highest monthly bill.

### median()

Finds the middle value after sorting the values.

Example: Median customer spending.

### std()

Calculates standard deviation to measure how much values vary.

Example: Variation in monthly customer spending.

### nunique()

Counts the number of unique values.

Example: Number of different services used by a customer.

## 4. Group-Based Aggregation

Group-based aggregation summarizes records for each group.

Example:

A telecom company has multiple transaction records for different customers.

We can group records by CustomerID and calculate:

* Total spending per customer.
* Average spending per customer.
* Number of transactions per customer.

In Pandas, we use:

`groupby()`

## 5. Multiple Aggregations

We can calculate several statistics at the same time.

Example:

`groupby("CustomerID").agg(["sum", "mean", "count", "max"])`

This creates multiple summarized features.

## 6. Rolling Aggregation

Rolling aggregation calculates statistics over a fixed moving window.

Example:

Average spending over the last 3 months.

It is useful for identifying recent changes in customer behaviour.

## 7. Cumulative Aggregation

Cumulative aggregation calculates a running total.

Example:

* January: ₹500
* February: ₹1,200
* March: ₹2,000

The cumulative spending increases as new transactions are added.

## 8. Customer-Level Aggregation

Customer-level aggregation combines information from multiple records into one row per customer.

Example features:

* Total_Spending
* Average_Spending
* Transaction_Count
* Maximum_Spending
* Minimum_Spending

These features can help identify high-value and low-activity customers.

## 9. Important Considerations

* Choose meaningful grouping columns.
* Understand what each aggregation represents.
* Handle missing values appropriately.
* Avoid aggregating information from future records.
* Use the correct observation period.
* Avoid including target information in input features.
* Check whether the aggregated feature provides useful information.

## Conclusion

Aggregation Feature Engineering converts detailed records into summarized features.

Functions such as sum, mean, count, minimum, maximum, and standard deviation help Machine Learning models understand customer-level patterns and business behaviour.


# Demonstration: Aggregation Features in a Telecom Business

## Business Scenario

A telecom company maintains transaction records for its customers.

Each customer may have multiple transactions.

The company wants to understand customer spending behaviour.

Consider the following transaction data:

| CustomerID | Transaction Amount |
| ---------- | -----------------: |
| 101        |                500 |
| 101        |                700 |
| 101        |                800 |
| 102        |                300 |
| 102        |                400 |
| 103        |                900 |
| 103        |              1,000 |

## Example 1: Total Spending

For Customer 101:

500 + 700 + 800 = 2,000

Feature: `Total_Spending`

Reason: Identifies the total amount spent by a customer.

## Example 2: Average Spending

For Customer 101:

2,000 / 3 = 666.67

Feature: `Average_Spending`

Reason: Identifies the customer's average transaction amount.

## Example 3: Transaction Count

Customer 101 has made 3 transactions.

Feature: `Transaction_Count`

Reason: Measures customer activity.

## Example 4: Maximum and Minimum Spending

Customer 101:

* Maximum spending = 800
* Minimum spending = 500

Features:

* `Maximum_Spending`
* `Minimum_Spending`

Reason: Helps understand spending range.

## Example 5: Spending Variation

Standard deviation measures how much transaction amounts vary.

Feature: `Spending_Std`

Reason: Helps identify customers with stable or changing spending patterns.

## Business Value

Aggregation features help a telecom company:

* Identify high-value customers.
* Understand average customer spending.
* Measure customer activity.
* Identify changes in spending behaviour.
* Compare customer groups.

## ML Value

These features may help Machine Learning models identify patterns associated with customer behaviour and churn.

**Important:** Aggregations must only use records available before or at the prediction date. Future transactions must never be included.


```python
import pandas as pd
import numpy as np

# ---------------------------------------
# 1. Create sample transaction dataset
# ---------------------------------------

data = {
    "CustomerID": [
        101, 101, 101,
        102, 102,
        103, 103, 103,
        104, 104
    ],

    "TransactionDate": [
        "2026-01-10",
        "2026-02-15",
        "2026-03-20",
        "2026-01-12",
        "2026-02-18",
        "2026-01-05",
        "2026-02-10",
        "2026-03-15",
        "2026-01-20",
        "2026-03-25"
    ],

    "Amount": [
        500, 700, 800,
        300, 400,
        900, 1000, 1100,
        600, 750
    ],

    "Service": [
        "Internet", "Mobile", "Internet",
        "Mobile", "Internet",
        "Internet", "Mobile", "Internet",
        "Mobile", "Internet"
    ]
}

df = pd.DataFrame(data)

df["TransactionDate"] = pd.to_datetime(
    df["TransactionDate"]
)

print("Original Transaction Dataset:")
print(df)

# ---------------------------------------
# 2. Group-based aggregation
# ---------------------------------------

customer_features = (
    df.groupby("CustomerID")
    .agg(
        Total_Spending=("Amount", "sum"),
        Average_Spending=("Amount", "mean"),
        Transaction_Count=("Amount", "count"),
        Maximum_Spending=("Amount", "max"),
        Minimum_Spending=("Amount", "min"),
        Median_Spending=("Amount", "median"),
        Spending_Std=("Amount", "std"),
        Unique_Services=("Service", "nunique")
    )
    .reset_index()
)

print("\nCustomer-Level Aggregation:")
print(customer_features)

# ---------------------------------------
# 3. Replace missing standard deviation
# ---------------------------------------

customer_features["Spending_Std"] = (
    customer_features["Spending_Std"]
    .fillna(0)
)

# ---------------------------------------
# 4. Calculate monthly aggregation
# ---------------------------------------

df["TransactionMonth"] = (
    df["TransactionDate"].dt.to_period("M")
)

monthly_features = (
    df.groupby(["CustomerID", "TransactionMonth"])
    .agg(
        Monthly_Spending=("Amount", "sum"),
        Monthly_Transactions=("Amount", "count")
    )
    .reset_index()
)

print("\nMonthly Aggregation:")
print(monthly_features)

# ---------------------------------------
# 5. Calculate rolling aggregation
# ---------------------------------------

customer_101 = (
    df[df["CustomerID"] == 101]
    .sort_values("TransactionDate")
    .copy()
)

customer_101["Rolling_Average_2"] = (
    customer_101["Amount"]
    .rolling(window=2, min_periods=1)
    .mean()
)

print("\nRolling Average for Customer 101:")
print(
    customer_101[
        [
            "CustomerID",
            "TransactionDate",
            "Amount",
            "Rolling_Average_2"
        ]
    ]
)

# ---------------------------------------
# 6. Calculate cumulative spending
# ---------------------------------------

customer_101["Cumulative_Spending"] = (
    customer_101["Amount"].cumsum()
)

print("\nCumulative Spending for Customer 101:")
print(
    customer_101[
        [
            "TransactionDate",
            "Amount",
            "Cumulative_Spending"
        ]
    ]
)

# ---------------------------------------
# 7. Merge aggregation features
# ---------------------------------------

final_dataset = customer_features.copy()

print("\nFinal Customer Feature Dataset:")
print(final_dataset)

print("\nFinal Dataset Shape:")
print(final_dataset.shape)

print("\nMissing Values:")
print(final_dataset.isnull().sum())
```

Their usefulness should be checked through statistical analysis and model performance.

## 4. Leakage Prevention

Aggregation features must use only information available at the prediction time.

For example:

If the company predicts churn on 1 April 2026, it should not include transactions from May or June 2026.

For historical Machine Learning datasets:

1. Define the observation period.
2. Filter records using the prediction date.
3. Calculate aggregations using only the permitted records.
4. Keep future target information separate from input features.

## 5. Final Conclusion

Aggregation Feature Engineering converts multiple transaction records into meaningful customer-level features.

Features such as total spending, average spending, transaction count, and unique services can help Machine Learning models understand customer behaviour.

However, every aggregation must have a clear business purpose and must be checked for future-data leakage before being used for training.
