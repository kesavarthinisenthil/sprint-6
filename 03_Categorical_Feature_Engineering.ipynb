# Categorical Feature Engineering

## 1. What is Categorical Feature Engineering?

Categorical Feature Engineering is the process of converting categorical information into a numerical format that Machine Learning algorithms can understand.

Categorical data represents groups, labels or categories rather than numerical measurements.

Examples:

* Gender
* Country
* Contract Type
* Payment Method
* Product Category
* Internet Service

Machine Learning algorithms generally require numerical inputs, so categorical variables must be appropriately encoded.

## 2. Types of Categorical Variables

### 2.1 Nominal Variables

Nominal variables represent categories without any natural order.

Examples:

* City: Chennai, Coimbatore, Madurai
* Payment Method: Credit Card, Cash, UPI
* Internet Service: DSL, Fiber Optic, No

There is no ranking between these categories.

Common technique: One-Hot Encoding.

### 2.2 Ordinal Variables

Ordinal variables represent categories that have a meaningful order or ranking.

Examples:

* Education: School, Bachelor's, Master's, PhD
* Satisfaction: Low, Medium, High
* Customer Rating: Poor, Average, Good, Excellent

These categories have a meaningful order.

Common technique: Ordinal Encoding.

### 2.3 Binary Variables

Binary variables contain only two possible categories.

Examples:

* Yes / No
* True / False
* Active / Inactive
* Churn / Not Churn

They can be represented using 0 and 1.

Example:

* Yes = 1
* No = 0

## 3. One-Hot Encoding

One-Hot Encoding converts each category into a separate binary column.

Example:

Original column:

| Payment Method |
| -------------- |
| UPI            |
| Cash           |
| Card           |

After One-Hot Encoding:

| Payment_UPI | Payment_Cash | Payment_Card |
| ----------: | -----------: | -----------: |
|           1 |            0 |            0 |
|           0 |            1 |            0 |
|           0 |            0 |            1 |

When to use:

* Nominal categorical variables.
* Categories without a natural order.
* Features with a manageable number of unique categories.

Limitation:
It can create too many columns when a feature contains many unique categories.

## 4. Ordinal Encoding

Ordinal Encoding converts ordered categories into numerical values while preserving their ranking.

Example:

| Satisfaction | Encoded Value |
| ------------ | ------------: |
| Low          |             1 |
| Medium       |             2 |
| High         |             3 |

When to use:

* Education levels.
* Customer satisfaction.
* Product ratings.
* Risk categories.

Important:
The assigned numerical values should represent a meaningful order.

## 5. Label Encoding

Label Encoding assigns a unique integer to each category.

Example:

| Contract |
|---|---:|
| Month-to-month | 0 |
| One year | 1 |
| Two year | 2 |

When to use:

* Encoding target labels.
* Certain categorical features when the encoding is appropriate for the model and category meaning.

Limitation:
For nominal input features, integer codes can introduce an artificial order that does not actually exist.

## 6. Frequency Encoding

Frequency Encoding replaces each category with the number of times it appears in the dataset.

Example:

| City       | Frequency |
| ---------- | --------: |
| Chennai    |         3 |
| Coimbatore |         2 |
| Madurai    |         1 |

When to use:

* High-cardinality categorical features.
* Features containing many unique categories.
* When category frequency may provide useful information.

Limitation:
Different categories may have the same frequency.

## 7. Target Encoding

Target Encoding replaces each category with a statistic calculated from the target variable.

For a binary classification problem, this is commonly the mean target value for each category.

Example:

| Contract       | Average Churn |
| -------------- | ------------: |
| Month-to-month |          0.40 |
| One year       |          0.15 |
| Two year       |          0.05 |

When to use:

* High-cardinality categorical variables.
* When categories have a meaningful relationship with the target.

Limitation:
It can cause target leakage and overfitting if implemented incorrectly.

Use cross-validation-based encoding or out-of-fold encoding, and fit the encoder only on training data.

## 8. Rare Category Grouping

Rare Category Grouping combines categories that occur very infrequently into a single category.

Example:

Before grouping:

* Chennai: 100
* Coimbatore: 80
* Madurai: 60
* Salem: 2
* Erode: 1

After grouping:

* Chennai: 100
* Coimbatore: 80
* Madurai: 60
* Other: 3

When to use:

* Features with many rare categories.
* Categories with very few observations.
* To reduce unnecessary complexity.

Limitation:
Rare categories may contain useful information, so grouping should be evaluated carefully.

## 9. Category Cardinality

Cardinality refers to the number of unique categories in a categorical feature.

Example:

| Feature     | Unique Categories | Cardinality |
| ----------- | ----------------- | ----------- |
| Gender      | 2                 | Low         |
| Contract    | 3                 | Low         |
| City        | 50                | Medium      |
| Customer ID | 10000             | High        |

High-cardinality features may create too many columns when One-Hot Encoding is used.

Possible techniques:

* Frequency Encoding
* Target Encoding
* Rare Category Grouping
* Feature Removal, when appropriate

## 10. Handling Unknown Categories

Unknown categories are new category values that were not present when the model was trained.

Example:

Training categories:

* UPI
* Cash
* Card

New prediction data:

* Net Banking

The encoder must handle this new category without producing an error.

In Scikit-learn, OneHotEncoder supports:

handle_unknown="ignore"

This prevents errors when previously unseen categories appear.

## 11. Combining Categories

Combining Categories means grouping related categories into a meaningful new category.

Example:

Original categories:

* Fiber Optic
* DSL
* Cable
* No Internet

Combined categories:

* Wired Internet
* No Internet

When to use:

* Reducing unnecessary category complexity.
* Grouping similar business categories.
* Creating meaningful customer segments.

The grouping should be based on domain knowledge.

## 12. Comparison of Encoding Techniques

| Technique              | Purpose                      | Suitable Data                    |
| ---------------------- | ---------------------------- | -------------------------------- |
| One-Hot Encoding       | Creates binary columns       | Nominal                          |
| Ordinal Encoding       | Preserves category order     | Ordinal                          |
| Label Encoding         | Assigns integer labels       | Target labels or suitable inputs |
| Frequency Encoding     | Uses category frequency      | High cardinality                 |
| Target Encoding        | Uses target statistics       | High cardinality                 |
| Rare Category Grouping | Groups infrequent categories | Rare categories                  |

## 13. Importance of Categorical Feature Engineering

Categorical Feature Engineering helps:

1. Convert text categories into numerical values.
2. Represent business categories meaningfully.
3. Reduce unnecessary complexity.
4. Handle high-cardinality features.
5. Improve Machine Learning model performance.
6. Preserve meaningful category relationships.

## Conclusion

Categorical Feature Engineering converts categorical information into useful numerical representations.

The encoding technique should be selected based on the type of categorical variable, number of unique categories, business meaning and Machine Learning algorithm.

An AI/ML Engineer must also handle unknown categories and prevent leakage during encoding.

## 2.5 Why is Categorical Feature Engineering Useful?

1. Converts text categories into numerical representations.
2. Helps Machine Learning algorithms process categorical information.
3. Preserves category order when ordinal encoding is appropriate.
4. Handles features with many unique categories.
5. Helps represent business information.
6. Supports better feature preparation.

## Conclusion

Categorical Feature Engineering is important because real-world datasets contain many text-based categories.

Choosing the correct encoding technique helps preserve useful information and prepare categorical variables for Machine Learning.


```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    OneHotEncoder,
    OrdinalEncoder,
    LabelEncoder
)

# 1. Create Sample Telecom Dataset

data = {
    "CustomerID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Contract": [
        "Month-to-month",
        "One year",
        "Two year",
        "Month-to-month",
        "One year",
        "Two year",
        "Month-to-month",
        "Month-to-month"
    ],
    "InternetService": [
        "Fiber optic",
        "DSL",
        "No",
        "Fiber optic",
        "DSL",
        "No",
        "Fiber optic",
        "Fiber optic"
    ],
    "PaymentMethod": [
        "Electronic check",
        "Mailed check",
        "Bank transfer",
        "Credit card",
        "Electronic check",
        "Mailed check",
        "Electronic check",
        "Credit card"
    ],
    "Satisfaction": [
        "High",
        "Medium",
        "Low",
        "High",
        "Medium",
        "Low",
        "High",
        "Medium"
    ],
    "PaperlessBilling": [
        "Yes", "No", "Yes", "No",
        "Yes", "No", "Yes", "Yes"
    ],
    "Churn": [
        "Yes", "No", "No", "Yes",
        "No", "No", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)


# 2. One-Hot Encoding

one_hot_df = pd.get_dummies(
    df["Contract"],
    prefix="Contract",
    dtype=int
)

df = pd.concat([df, one_hot_df], axis=1)


# 3. Ordinal Encoding

ordinal_encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["Satisfaction_Encoded"] = ordinal_encoder.fit_transform(
    df[["Satisfaction"]]
).astype(int)


# 4. Label Encoding

label_encoder = LabelEncoder()

df["Churn_Encoded"] = label_encoder.fit_transform(
    df["Churn"]
)


# 5. Frequency Encoding

frequency_map = df["PaymentMethod"].value_counts()

df["Payment_Frequency"] = (
    df["PaymentMethod"].map(frequency_map)
)


# 6. Target Encoding Demonstration

df["Churn_Numeric"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})

target_map = df.groupby("InternetService")[
    "Churn_Numeric"
].mean()

df["Internet_Target_Encoded"] = (
    df["InternetService"].map(target_map)
)


# 7. Rare Category Grouping

payment_counts = df["PaymentMethod"].value_counts()

rare_categories = payment_counts[
    payment_counts < 2
].index

df["Payment_Grouped"] = df["PaymentMethod"].replace(
    rare_categories,
    "Other"
)


# 8. Category Cardinality

cardinality = df[
    ["Contract", "InternetService", "PaymentMethod"]
].nunique()

print("\nCategory Cardinality:")
print(cardinality)


# 9. Combining Categories

df["Service_Category"] = df["InternetService"].replace({
    "DSL": "Wired Internet",
    "Fiber optic": "Wired Internet",
    "No": "No Internet"
})


# 10. Unknown Category Handling using OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoded_array = encoder.fit_transform(
    df[["InternetService"]]
)

encoded_columns = encoder.get_feature_names_out(
    ["InternetService"]
)

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoded_columns,
    index=df.index
)

print("\nOne-Hot Encoding with Scikit-learn:")
display(encoded_df)


# Final Dataset

print("\nFinal Engineered Dataset:")
display(df)

print("\nFinal Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())
```
