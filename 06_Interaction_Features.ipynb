# Interaction Feature Engineering

## 1. What are Interaction Features?

Interaction Feature Engineering is the process of combining two or more existing features to create a new feature that represents their relationship.

Sometimes, one feature alone cannot explain customer behaviour.

For example:
- Monthly Charges
- Customer Tenure

Individually, these features provide separate information.

When we multiply them:

Estimated Customer Value = Monthly Charges × Customer Tenure

The new feature represents the estimated total charges over the customer's tenure, assuming the monthly charge remains constant.

## 2. Why are Interaction Features Important?

Interaction features help Machine Learning models understand how multiple variables work together.

They can help:
- Identify relationships between features.
- Capture combined effects.
- Improve predictive performance.
- Represent business behaviour.
- Discover hidden patterns.

## 3. Multiplication Interaction

Multiplication combines two numerical features.

Formula:

Interaction Feature = Feature A × Feature B

Example:

Monthly Charges = ₹500  
Tenure = 12 months

Estimated Total Charges = 500 × 12 = ₹6,000

**ML Use:** Represents the combined effect of monthly charges and customer tenure.

## 4. Division Interaction

Division creates a ratio between two features.

Formula:

Ratio Feature = Feature A / Feature B

Example:

Total Charges = ₹12,000  
Tenure = 24 months

Average Monthly Charges = 12,000 / 24 = ₹500

**ML Use:** Helps compare values relative to another feature.

Important: Handle zero values in the denominator to avoid division errors.

## 5. Addition Interaction

Addition combines two or more numerical features.

Formula:

Combined Charges = Internet Charges + Phone Charges

**ML Use:** Represents the total cost of multiple services.

## 6. Subtraction Interaction

Subtraction measures the difference between two features.

Formula:

Charge Difference = Internet Charges - Phone Charges

**ML Use:** Helps identify differences in service spending.

## 7. Polynomial Interaction

Polynomial interaction features are created by multiplying features or raising them to powers.

Example:

Feature A × Feature B

Feature A²

Feature B²

**ML Use:** Helps capture nonlinear relationships when appropriate.

## 8. Categorical-Numerical Interaction

We can combine a categorical feature with a numerical feature.

Example:

Contract Type = Two Year  
Monthly Charges = ₹600

The combined information can represent monthly spending for customers under a particular contract category.

This can be implemented using one-hot encoding followed by interaction with numerical features.

**ML Use:** Helps identify whether numerical behaviour differs across customer categories.

## 9. Feature Explosion

Creating too many interaction features can increase the number of input variables.

This is called feature explosion.

Problems include:
- Increased model complexity.
- More memory usage.
- Overfitting.
- Redundant features.
- Difficult interpretation.

Therefore, create only meaningful interactions.

## 10. Important Considerations

- Understand the business relationship before combining features.
- Check whether the new feature provides useful information.
- Avoid division by zero.
- Check for highly correlated features.
- Avoid creating too many interactions.
- Prevent future-data leakage.
- Evaluate the feature using model performance.

## Conclusion

Interaction Feature Engineering creates meaningful combinations of existing features.

Multiplication, division, addition, subtraction, and polynomial operations help Machine Learning models understand relationships between variables.

Every interaction feature should have a clear business purpose and should be evaluated before being retained.

# Demonstration: Interaction Features in a Telecom Business

## Business Scenario

A telecom company wants to understand how customer spending and service usage are related.

The company has information such as:
- Monthly Charges
- Customer Tenure
- Internet Charges
- Phone Charges
- Contract Type
- Data Usage

We can combine these features to create meaningful interactions.

## Example 1: Monthly Charges × Tenure

Monthly Charges = ₹500  
Tenure = 12 months

Estimated Total Charges = 500 × 12 = ₹6,000

Feature Name: `Estimated_Total_Charges`

Reason: Represents estimated customer spending over the tenure period.

## Example 2: Internet Charges + Phone Charges

Internet Charges = ₹400  
Phone Charges = ₹100

Combined Charges = ₹500

Feature Name: `Combined_Service_Charges`

Reason: Represents the combined cost of two services.

## Example 3: Internet Charges / Phone Charges

Internet Charges = ₹400  
Phone Charges = ₹200

Ratio = 400 / 200 = 2

Feature Name: `Internet_Phone_Ratio`

Reason: Shows the relative cost of internet service compared with phone service.

## Example 4: Monthly Charges - Internet Charges

Monthly Charges = ₹600  
Internet Charges = ₹400

Other Service Charges = ₹200

Feature Name: `Other_Service_Charges`

Reason: Estimates the remaining charges after subtracting internet charges, assuming the source columns are defined consistently.

## Example 5: Contract Type × Monthly Charges

Contract Type = Two Year  
Monthly Charges = ₹600

We can create an encoded contract feature and multiply it by monthly charges.

Reason: Helps identify spending patterns associated with different contract types.

## Business Value

Interaction features help the telecom company:
- Understand combined customer spending.
- Compare service costs.
- Identify customer spending patterns.
- Understand differences between contract groups.

## ML Value

These features may help Machine Learning models capture relationships that are not obvious when examining individual features separately.

However, interaction features should be tested because some combinations may not improve prediction quality.

```python
import pandas as pd
import numpy as np

# ---------------------------------------
# 1. Create sample telecom dataset
# ---------------------------------------

data = {
    "CustomerID": [101, 102, 103, 104, 105],

    "MonthlyCharges": [500, 600, 450, 700, 550],

    "Tenure": [12, 24, 6, 36, 18],

    "InternetCharges": [300, 400, 250, 450, 350],

    "PhoneCharges": [200, 200, 200, 250, 200],

    "DataUsageGB": [50, 80, 30, 100, 65],

    "Contract": [
        "Monthly",
        "Yearly",
        "Monthly",
        "Two Year",
        "Yearly"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# ---------------------------------------
# 2. Multiplication interaction
# ---------------------------------------

df["Estimated_Total_Charges"] = (
    df["MonthlyCharges"] * df["Tenure"]
)

# ---------------------------------------
# 3. Addition interaction
# ---------------------------------------

df["Combined_Service_Charges"] = (
    df["InternetCharges"] + df["PhoneCharges"]
)

# ---------------------------------------
# 4. Division interaction
# ---------------------------------------

df["Internet_Phone_Ratio"] = np.where(
    df["PhoneCharges"] != 0,
    df["InternetCharges"] / df["PhoneCharges"],
    np.nan
)

# ---------------------------------------
# 5. Subtraction interaction
# ---------------------------------------

df["Other_Service_Charges"] = (
    df["MonthlyCharges"] - df["InternetCharges"]
)

# ---------------------------------------
# 6. Numerical polynomial interaction
# ---------------------------------------

df["MonthlyCharges_Squared"] = (
    df["MonthlyCharges"] ** 2
)

df["DataUsage_Tenure_Interaction"] = (
    df["DataUsageGB"] * df["Tenure"]
)

# ---------------------------------------
# 7. Categorical-numerical interaction
# ---------------------------------------

contract_encoded = pd.get_dummies(
    df["Contract"],
    prefix="Contract",
    dtype=int
)

df = pd.concat(
    [df, contract_encoded],
    axis=1
)

df["Yearly_Contract_Charges"] = (
    df["Contract_Yearly"] * df["MonthlyCharges"]
)

df["TwoYear_Contract_Charges"] = (
    df["Contract_Two Year"] * df["MonthlyCharges"]
)

# ---------------------------------------
# 8. Display engineered dataset
# ---------------------------------------

print("\nDataset After Interaction Feature Engineering:")
print(df)

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nNew Interaction Features:")

new_features = [
    "Estimated_Total_Charges",
    "Combined_Service_Charges",
    "Internet_Phone_Ratio",
    "Other_Service_Charges",
    "MonthlyCharges_Squared",
    "DataUsage_Tenure_Interaction",
    "Yearly_Contract_Charges",
    "TwoYear_Contract_Charges"
]

print(df[new_features])
```

## 2. Important Observations

- Multiplication captures combined numerical effects.
- Division captures relative relationships.
- Addition and subtraction represent combined or remaining values.
- Categorical-numerical interactions capture differences between groups.
- Polynomial features may help capture nonlinear patterns.

## 3. Feature Selection Decisions

- Estimated_Total_Charges
- Combined_Service_Charges

These features have clear potential business meanings.

### Features Requiring Further Analysis

- Other_Service_Charges
- MonthlyCharges_Squared
- Yearly_Contract_Charges
- TwoYear_Contract_Charges

Their value should be tested using feature selection and model evaluation.

## 4. Leakage Prevention

Interaction features must only use information available at prediction time.

For example:

If monthly charges are calculated using future billing information, combining them with tenure can introduce future information into the model.

Always:
1. Check the availability of source columns.
2. Confirm that source values are available at prediction time.
3. Avoid using future information.
4. Separate target variables from input features.
5. Validate interactions using training and testing data.

## 5. Final Conclusion

Interaction Feature Engineering helps Machine Learning models understand relationships between multiple features.

Features such as estimated total charges, combined service charges, and usage-tenure interactions can provide useful business information.

However, creating too many interactions can increase model complexity and overfitting.

Therefore, interaction features should be selected based on business meaning, leakage risk, and actual model performance.
